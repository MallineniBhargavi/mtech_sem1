```python
# 01. Flower

class Flower:
    def __init__(self,name: str ="unknown", petals: int=0, price: float=0.0):
        self._name = str(name)
        self._petals = str(petals)
        self._price = str(price)

    def set_name(self, name: str)-> None:
        self._name = str(name)

    def set_petals(self, petals: int)-> None:
        if petals < 0:
            raise ValueError("Number of petals cannot be negative.") 
        self._petals = int(petals)

    def set_price(self, price: float)-> None:
        if price < 0:
            raise ValueError("price cannot be negative.")
        self._price = float(price)

    def get_name(self) -> str:
        return self._name

    def get_petals(self) -> int:
        return self._petals # type: ignore
        
    def get_price(self) -> float:
        return self._price # type: ignore

    def __str__(self) -> str:
        return f"flower (name= '{self._name}', petals= {self._petals}, price= ${self._price:.2f})"
    
def main():
    rose = Flower("rose",24,5.99)
    print("Initial flower details:")
    print("rose")
    print("\n using getters:")
    print(f"Name : {rose.get_name()}")
    print(f"petals : {rose.get_petals()}")
    print(f"price : {rose.get_price()}")
    print("\n updating flower using setters....")
    rose.set_name("Red Rose")
    rose.set_petals(30)
    rose.set_price(7.49)
    print("\n updated flower details:")
    print(rose)

if __name__=="__main__":
    main()

```

```python
# 02: Polygon_hierarchy 

from abc import ABC, abstractmethod
import math

class Polygon(ABC):
    def area(self) -> float:
        raise NotImplementedError
    def perimeter(self) -> float:
        raise NotImplementedError
    def describe(self) -> str:
        return(f"{self.__class__.__name__}:"f"area = {self.area():.2f},perimeter = {self.perimeter():.2f}")
    
class Traingle(Polygon):
    def __init__(self, side_a: float, side_b: float, side_c: float) :
        if(side_a + side_b <= side_c or side_b + side_c <= side_a or side_a + side_c <= side_b):
            raise ValueError("The given Sides do not form a vaild traingle.")
        self.side_a = side_a
        self.side_b = side_b
        self.side_c = side_c
    def area(self) -> float:
        s = self.perimeter()/2
        return math.sqrt(s * (s - self.side_a) * (s- self.side_b) * (s - self.side_c))
    def perimeter(self) -> float:
        return self.side_a + self.side_b + self.side_c
    
class Quadrilateral(Polygon):
    def __init__(self, length:float, width: float):
        self.length = length
        self.width = width
    def area(self) -> float:
        return self.length * self.width
    def perimeter(self) -> float:
        return 2 * (self.length + self.width)
class Pentagon(Polygon):
    def __init__(self, side: float):
        self.side = side
    def area(self) -> float:
        return 0.25 * math.sqrt(5 * (5+2 * math.sqrt(5))) * (self.side ** 2)
    def perimeter(self) -> float:
        return 5 * self.side

def _read_float(prompt: str) -> float:
    while True:
        try:
            return float (input(prompt))
        except ValueError:
            print("please enter a valid Number.")

def create_polygon_interactively() -> Polygon:
    print("\nChoose a polygon type:")
    print(" 1. Triangle (3 sides)")
    print(" 2. Quadrilateral (length X width)")
    print(" 3. Pentagon (regular,1 side)")
    choice = input(" Enter choice (1/2/3):").strip()
    if choice == "1":
        a = _read_float("Side A:")
        b = _read_float("Side B:")
        c = _read_float("Side C:")
        return Traingle(a,b,c) 
    elif choice == "2":
        length = _read_float("length: ")
        width = _read_float("width:")
        return Quadrilateral (length, width)
    elif choice == "3":
        side = _read_float("side length:")
        return Pentagon(side)
    else:
        raise ValueError("Invalid choice. please choose 1,2 or 3.")
def demo():
    print("=== Demo mode (sample polygons) ===")
    shapes = [Traingle(3,4,5), Quadrilateral(6,4), Pentagon(5)]
    for shape in shapes:
        print(shape.describe())

def main():
    demo()
    print("\n===Interactive Mode ===")
    answer = input(" would you like to create your own polygon(s)? (y/n) : ").strip().lower()
    while answer == "y":
        try:
            polygon = create_polygon_interactively()
            print(polygon.describe())
        except ValueError as e:
            print(f"Error: {e}")
        answer = input("\n Create another polygon? (y/n):").strip().lower()
    print("Good Bye!")
if __name__== "__main__":
    main()
```

```python
# 3: Method Overloading and Method Overriding

import math

class Shape:
    def area(self, *dimensions):
        if len(dimensions) == 1:
            radius = dimensions[0]
            return math.pi * radius ** 2
        elif len(dimensions) == 2:
            length, width = dimensions
            return length * width
        elif len(dimensions) == 3:
            a, b, c = dimensions
            s = (a + b + c) / 2
            return math.sqrt(s * (s - a) * (s - b) * (s - c))
        else:
            raise ValueError("area() supports 1, 2, or 3 dimensions only.")
        
    def describe(self) -> str:
        """Default description; subclasses OVERRIDE this method."""
        return "I am a generic shape."
    
class Circle(Shape):
    """A circle - overrides describe() from Shape."""
    def __init__(self, radius: float):
        self.radius = radius
    def describe(self) -> str:
        """Override: provide a circle-specific description."""
        return f"I am a Circle with radius {self.radius} and area {self.area(self.radius):.2f}."

class Rectangle(Shape):
    """A rectangle - overrides describe() from Shape."""
    def __init__(self, length: float, width: float):
        self.length = length
        self.width = width
    def describe(self) -> str:
        """Override: provide a rectangle-specific description."""
        return (f"I am a Rectangle {self.length}x{self.width} "
                f"with area {self.area(self.length, self.width):.2f}.")

def main():
    """Demonstrate both overloading (via *args) and overriding (via subclasses)."""
    print("=== Method Overloading Demo (single area() method, variable args) ===")
    shape = Shape()
    print(f"area(5) -> circle area : {shape.area(5):.2f}")
    print(f"area(4, 6) -> rectangle area : {shape.area(4, 6):.2f}")
    print(f"area(3, 4, 5) -> triangle area : {shape.area(3, 4, 5):.2f}")
    print("\n=== Method Overriding Demo (subclasses override describe()) ===")
    shapes = [Shape(), Circle(5), Rectangle(4, 6)]
    for s in shapes:
        # The SAME method call, describe(), produces different output
        # depending on the actual (runtime) type of the object -
        # this is polymorphism enabled by method overriding.
        print(s.describe())

if __name__ == "__main__":
    main()
```

```python
# 04. Comprehensions

def list_comprehension_demo(numbers):
    
    return [n ** 2 for n in numbers]

def dict_comprehension_demo(numbers):
    """Return a dict mapping each number to its square using a
    dictionary comprehension.
    Args:
    numbers (list[int]): Input numbers.
    Returns:
    dict[int, int]: Mapping of number -> square.
    """
    return {n: n ** 2 for n in numbers}

def set_comprehension_demo(numbers):
    """Return the set of remainders (mod 3) using a set comprehension.
    (A set automatically removes duplicate values.)
    Args:
    numbers (list[int]): Input numbers.
    Returns:
    set[int]: Unique remainders when each number is divided by 3.
    """
    return {n % 3 for n in numbers}

def generator_comprehension_demo(numbers):
    """Return a generator that lazily yields the cube of each number.
    Args:
    numbers (list[int]): Input numbers.
    Returns:
    generator: A generator object; values are computed on demand
    when iterated (e.g., with next() or a for-loop).
    """
    return (n ** 3 for n in numbers)

def main():
    """Demonstrate all four comprehension types on the same input data."""
    numbers = [1, 2, 3, 4, 5, 6]
    print(f"Input numbers: {numbers}\n")
    # a) List comprehension
    squares = list_comprehension_demo(numbers)
    print(f"a) List Comprehension (squares) : {squares}")
    # b) Dictionary comprehension
    square_map = dict_comprehension_demo(numbers)
    print(f"b) Dictionary Comprehension (n -> n^2): {square_map}")
    # c) Set comprehension
    remainders = set_comprehension_demo(numbers)
    print(f"c) Set Comprehension (n % 3, unique) : {remainders}")
    # d) Generator comprehension
    cubes_gen = generator_comprehension_demo(numbers)
    print("d) Generator Comprehension (cubes) : ", end="")
    # A generator can only be iterated once; convert to a list to display it
    # (in real use, you'd typically loop over it directly to save memory).
    print(list(cubes_gen))

if __name__ == "__main__":
    main()
```
